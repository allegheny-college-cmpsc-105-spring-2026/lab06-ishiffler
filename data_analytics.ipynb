```python
# data_analytics.csv

# : update the column names to match the names you have in your csv file for data_analytics

# create a pandas dataframe
# Temporarily removing 'names' and 'skiprows' to let pandas infer the columns.
# Please inspect the output of dfa.head() to verify the column names and data.
# If the first row of your CSV contains headers, you might not need 'skiprows=1'.
# If your CSV does not have a header, you might need to add 'header=None' and
# then manually specify the column names using the 'names' parameter.
dfa = pd.read_csv(
    '/content/data_analytics.csv',
    sep=',',
    comment='#'
    # names = ['TODO', 'Replace', 'Your', 'Column names', 'Here', 'separated by commas as needed'],
    # skiprows=1
)

# Display the first few rows of the DataFrame
print(dfa.head())
```

```python
# data_survey.csv

# : update the column names to match the names you have in your csv file for data_survey

# create a pandas dataframe
dfs = pd.read_csv(
    '/content/data_survey.csv',
    sep=','
    # names = ['TODO', 'Replace', 'Your', 'Column names', 'Here', 'separated by commas as needed'],
    # skiprows=1
)

# Display the first few rows of the DataFrame
print(dfs.head())
```

```python
# : how many rows and columns are in the dataframe dfa? 4 rows and 8 columns
# : how many rows and columns are in the dataframe dfs? 7 rows and 6 columns
```

```python
# Analytics

# : Do one or more explorations with dfa
# no interaction is required

# make a blank canvas
plt.figure(figsize=(10, 6))

# make a title starting with 'Figure 1:' Fill in a relevant title
plt.title('Figure 1: Views by Page Path and Screen Class')

# plot
plt.bar(dfa['Page path and screen class'], dfa['Views'])
plt.xlabel('Page Path and Screen Class')
plt.ylabel('Views')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()

# reveal everything
plt.show()
```

```python
# Survey

# : Do one or more explorations with dfs
# no interaction is required

# make a blank canvas
plt.figure(figsize=(10, 6))

# make a title starting with 'Figure 2:' Fill in a relevant title
plt.title('Figure 2: Distribution of Favorite Sports')

# plot
dfs['What is your favorite sport?'].value_counts().plot(kind='bar')
plt.xlabel('Favorite Sport')
plt.ylabel('Number of Responses')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()

# reveal everything
plt.show()
```
